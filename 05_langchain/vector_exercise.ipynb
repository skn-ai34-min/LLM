```python
# Cosine Similarity

from sklearn.metrics.pairwise import cosine_similarity
import numpy as np

vector_a = np.array([[1, 0, -1]])
vector_b = np.array([[0, 1, -1]])

similarity = cosine_similarity(vector_a,vector_b)

similarity[0][0]
```

```python
# Faiss (L2 거리 기반 인덱싱 기법 적용)
import faiss
import numpy as np

# 1. Define vector dimension (e.g., 128)
dimension = 128
# 2. Create a standard flat L2 index
data = np.random.random((100, dimension)).astype('float32')
query = np.random.random((1, dimension)).astype('float32')
index = faiss.IndexFlatL2(dimension)
index.add(data)

k = 5
distances, indices = index.search(query, k)
print(indices)
print(distances)

```
