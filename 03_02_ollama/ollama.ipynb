```python
import ollama

response1 = ollama.chat(model='llama3', messages=[{'role': 'user', 'content': '안녕하세요!'}])
for i in response1:
    print(i)
```

```python
def ollama_chat(input_text):
    return ollama.chat(
        model='llama3',
        messages=[{'role': 'user', 'content': input_text}
])
```

```python
response1.message.content
```

```python
response2 = ollama_chat('파이썬으로 퀵소트를 구현해줘.')
for i in response2:
    print(i)
```

```python
response3 = ollama_chat(f'다음 텍스트를 한글 설명으로 바꿔줘: {response2.message.content}')
print(response3)
```
