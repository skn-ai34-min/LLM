## State
agent간에 공유될 정보를 State 개체(dict)로써 관리

```python
from typing import TypedDict, Annotated, List
from langgraph.graph.message import add_messages

class State(TypedDict):
    messages: Annotated[List, add_messages] # messages 필드가 add_messages 규칙으로 누적 저장
```

## 챗봇 노드
- Node(Agent) function
- Edge

```python
from langchain.chat_models import init_chat_model

llm = init_chat_model('gpt-5.6-luna')
# State의 messages를 입력으로 받아, 응답 메시지를 추가해주는 노드 함수
def chatbot(state: State):
    response = llm.invoke(state['messages']) # 누적된 메시지 리스트로 LLM 호출
    return {'messages': [response]} # add_messages 규칙에 의해 기존 messages에 응답이 누적됨
```

## 그래프 설계 / 컴파일

```python
from langgraph.graph import StateGraph, START, END

workflow = StateGraph(State) # State 스키마 기반 그래프 설계 객체 생성

workflow.add_edge(START, 'chatbot') # START -> chatbot
workflow.add_node('chatbot', chatbot)     # chatbot
workflow.add_edge('chatbot',END)    # chatbot -> END

graph = workflow.compile() # 설계된 workflow를 실행 가능한 그래프로 컴파일
print(type(graph))

graph
```

```python
user_query = '랭그래프가 뭐죠?'

# messages는 add_messages 규칙에 따라 계속해서 누적됨
state = {
    'messages': [
        ('system', '당신은 현명한 IT전문가 입니다.'),
        ('human', user_query)
    ]
}

response = graph.invoke(state) # 그래프 실행(START -> chatbot -> END)
print(response)
```

```python
print(response['messages'][-1].content) # 누적된 messages 중 마지막(모델 응답) 내용 출력
```

LangChain만으로도 “한 번 호출해서 답하는” RAG/챗봇은 충분히 만들 수 있다.
그러나 LangGraph가 따로 존재하는 이유는 LLM 앱이 커질수록 ‘흐름(상태, 분기, 반복, 복구)’을 안정적으로 관리해야만 하기 때문이다.

- 상태(State) 누적이 표준화됨
    - 대화/중간결과/도구결과를 State에 계속 쌓고, 노드가 그걸 읽고/추가하는 구조라서 파이프라인이 커져도 덜 꼬임.
- 분기/루프/조건 실행이 쉬움
    - “의도 분류 → (검색/계산/상담) 분기”, “답변이 부족하면 재검색 루프” 같은 제어흐름을 코드로 깔끔하게 표현.
- 멀티스텝 에이전트에 강함
    - 계획→도구호출→검증→최종답 같은 여러 단계를 노드로 쪼개 디버깅/테스트가 쉬움.
- 에러 복구/재시도 설계가 쉬움
    - 특정 노드 실패 시 재시도/대체 경로 등 운영 관점의 안정성을 넣기 좋음.
- 테스트/관찰(Observability) 포인트가 명확
    - 각 노드 입출력이 분리되니까 “어디서 망가졌는지” 추적이 쉬움(특히 LangSmith랑 붙일 때).

결론적으로
- LangChain = 구성요소(모델/프롬프트/리트리버/툴) 중심,
- LangGraph = 그 구성요소들을 ‘상태를 가진 워크플로우’로 운영하기 위한 프레임워크.
